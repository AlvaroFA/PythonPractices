### Ejercicio 1. Suma, recuento y media de una lista

Calcular, **sin ninguna biblioteca**, la suma de las duraciones, el número de películas y la duración media, redondeada a un decimal.

> **Ayuda:** las funciones integradas [`sum`](https://docs.python.org/es/3/library/functions.html#sum), [`len`](https://docs.python.org/es/3/library/functions.html#len) y [`round`](https://docs.python.org/es/3/library/functions.html#round). Las listas se explican en el [tutorial de Python](https://docs.python.org/es/3/tutorial/introduction.html#lists).

**Datos de partida.** Duración, en minutos, de cinco películas de [Agnès Varda](https://es.wikipedia.org/wiki/Agn%C3%A8s_Varda): *Cleo de 5 a 7* (1962), *La felicidad* (1965), *Sin techo ni ley* (1985), *Los espigadores y la espigadora* (2000) y *Caras y lugares* (2017).

````python
duraciones = [90, 80, 105, 82, 94]
````

**Resultado esperado.**


````python
# Escribe aquí tu solución y compara el resultado con el esperado
````

#### Solución del ejercicio 1

### Ejercicio 2. Recorrer una lista con una condición

Contar cuántas películas **duran más de 85 minutos**, primero con un bucle `for` y un `if`, y después en una sola línea con una **lista por comprensión**. Mostrar también la lista de esas duraciones.

> **Ayuda:** las sentencias [`for` e `if`](https://docs.python.org/es/3/tutorial/controlflow.html) y las [listas por comprensión](https://docs.python.org/es/3/tutorial/datastructures.html#list-comprehensions).

**Datos de partida.** Las mismas cinco duraciones de Agnès Varda.

````python
duraciones = [90, 80, 105, 82, 94]
````

**Resultado esperado.**

```text
Con bucle: 3
Por comprensión: 3 [90, 105, 94]
```

````python
# Escribe aquí tu solución y compara el resultado con el esperado
````

#### Solución del ejercicio 2

````python
contador = 0
for d in duraciones:
    if d > 85:
        contador += 1
print("Con bucle:", contador)

largas = [d for d in duraciones if d > 85]
print("Por comprensión:", len(largas), largas)
````

Las dos formas cuentan lo mismo. La lista por comprensión se lee casi como la frase en castellano: *las duraciones `d` de la lista, si `d` es mayor que 85*. En pandas no hará falta ninguna de las dos, porque las condiciones se aplican a toda la columna de una vez (ejercicio 7).

### Ejercicio 3. Operaciones con un array de NumPy

Convertir la lista en un **array de NumPy**, pasar todas las duraciones a **horas** con una sola operación (redondeadas a dos decimales) y obtener la media, el máximo y la **posición** del máximo. Con esa posición, mostrar el título de la película más larga.

> **Ayuda:** [`np.array`](https://numpy.org/doc/stable/reference/generated/numpy.array.html), [`np.mean`](https://numpy.org/doc/stable/reference/generated/numpy.mean.html), [`np.max`](https://numpy.org/doc/stable/reference/generated/numpy.max.html), [`np.argmax`](https://numpy.org/doc/stable/reference/generated/numpy.argmax.html) y [`np.round`](https://numpy.org/doc/stable/reference/generated/numpy.round.html). Recuerda que en Python las posiciones empiezan en 0.

**Datos de partida.** Las duraciones de Agnès Varda y sus títulos, en el mismo orden.

````python
import numpy as np

titulos = ["Cleo de 5 a 7", "La felicidad", "Sin techo ni ley",
           "Los espigadores y la espigadora", "Caras y lugares"]
duraciones = [90, 80, 105, 82, 94]
````

**Resultado esperado.**

```text
En horas: [1.5  1.33 1.75 1.37 1.57]
Media: 90.2   Máximo: 105   Posición del máximo: 2
La más larga: Sin techo ni ley
```

````python
# Escribe aquí tu solución y compara el resultado con el esperado
````

#### Solución del ejercicio 3

````python
d = np.array(duraciones)
print("En horas:", np.round(d / 60, 2))
print("Media:", d.mean(), "  Máximo:", d.max(), "  Posición del máximo:", d.argmax())
print("La más larga:", titulos[d.argmax()])
````

Con una lista, `duraciones / 60` daría un error; con un array, la división se aplica **a todos los elementos a la vez** (operación vectorizada). Esa es la razón de ser de NumPy, y pandas la hereda: cada columna de un `DataFrame` guarda sus datos en un array.

### Ejercicio 4. Una Series con índice

Crear una **`Series`** con el año de nacimiento de cinco compositoras, usando sus nombres como **índice**. Mostrar el año de Clara Schumann (por su etiqueta) y las compositoras nacidas **después de 1800**.

> **Ayuda:** [`pd.Series`](https://pandas.pydata.org/docs/reference/api/pandas.Series.html) con el parámetro `index`. Una condición entre corchetes filtra los elementos. Repaso en el apartado 1 del anexo 3.

**Datos de partida.** Nombres y años de nacimiento de [Hildegard von Bingen](https://es.wikipedia.org/wiki/Hildegarda_de_Bingen), [Barbara Strozzi](https://es.wikipedia.org/wiki/Barbara_Strozzi), [Fanny Mendelssohn](https://es.wikipedia.org/wiki/Fanny_Mendelssohn), [Clara Schumann](https://es.wikipedia.org/wiki/Clara_Schumann) y [Lili Boulanger](https://es.wikipedia.org/wiki/Lili_Boulanger).

````python
import pandas as pd

nombres = ["Hildegard von Bingen", "Barbara Strozzi", "Fanny Mendelssohn",
           "Clara Schumann", "Lili Boulanger"]
anios = [1098, 1619, 1805, 1819, 1893]
````

**Resultado esperado.**

```text

Fanny Mendelssohn    1805
Clara Schumann       1819
Lili Boulanger       1893
dtype: int64
```

````python
# Escribe aquí tu solución y compara el resultado con el esperado
````

#### Solución del ejercicio 4

````python
nacimiento = pd.Series(anios, index=nombres)
print("Clara Schumann:", nacimiento["Clara Schumann"])
print(nacimiento[nacimiento > 1800])
````

El índice sustituye a las posiciones 0, 1, 2... por etiquetas con significado, de modo que el dato se pide por su nombre. La condición `nacimiento > 1800` devuelve una `Series` de `True` y `False`, y al ponerla entre corchetes solo se conservan los elementos marcados con `True`.

### Ejercicio 7. Filtrar filas con condiciones

Mostrar el título y el año de las películas **estrenadas antes de 1960 o de más de 100 minutos**. Después, contar cuántas duran **entre 75 y 95 minutos**, ambos incluidos.

> **Ayuda:** las condiciones se combinan con `&` (y) y `|` (o), cada una entre paréntesis. [`Series.between`](https://pandas.pydata.org/docs/reference/api/pandas.Series.between.html) comprueba si un valor está en un intervalo. Repaso en el apartado 2 del anexo 3.

**Datos de partida.** La misma tabla de películas.

````python
import pandas as pd

peliculas = pd.DataFrame({
    "titulo": ["Baila, muchacha, baila", "El autoestopista", "Cleo de 5 a 7",
               "Las margaritas", "La ascensión"],
    "anio": [1940, 1953, 1962, 1966, 1977],
    "duracion": [90, 71, 90, 76, 111],
})
````

**Resultado esperado.**

```text
                   titulo  anio
0  Baila, muchacha, baila  1940
1        El autoestopista  1953
4            La ascensión  1977
Entre 75 y 95 minutos: 3
```

````python
# Escribe aquí tu solución y compara el resultado con el esperado
````

#### Solución del ejercicio 7

````python
seleccion = peliculas[(peliculas["anio"] < 1960) | (peliculas["duracion"] > 100)]
print(seleccion[["titulo", "anio"]])

entre = peliculas["duracion"].between(75, 95)
print("Entre 75 y 95 minutos:", entre.sum())
````

Sin los paréntesis, Python evaluaría primero `1960 | peliculas["duracion"]` y daría un error. Para contar se suma la `Series` de `True` y `False`: cada `True` vale 1 y cada `False`, 0.

### Ejercicio 9. Columnas calculadas

Añadir a la tabla dos columnas: `siglo`, calculada a partir del año (1929 pertenece al siglo 20), y `paginas_por_prestamo`, redondeada a un decimal. Mostrar el título y las dos columnas nuevas.

> **Ayuda:** la división entera `//` de Python y [`Series.round`](https://pandas.pydata.org/docs/reference/api/pandas.Series.round.html). Las operaciones entre columnas se aplican fila a fila sin escribir bucles. Repaso en el apartado 2 del anexo 3.

**Datos de partida.** Año de publicación, páginas y préstamos de cuatro libros (datos ilustrativos).

````python
import pandas as pd

libros = pd.DataFrame({
    "titulo": ["Una habitación propia", "Frankenstein", "Los pazos de Ulloa", "Nada"],
    "anio": [1929, 1818, 1886, 1945],
    "paginas": [160, 280, 384, 296],
    "prestamos": [34, 41, 18, 26],
})
````

**Resultado esperado.**

```text
                  titulo  siglo  paginas_por_prestamo
```

#### Solución del ejercicio 9

````python
libros["siglo"] = libros["anio"] // 100 + 1
libros["paginas_por_prestamo"] = (libros["paginas"] / libros["prestamos"]).round(1)
print(libros[["titulo", "siglo", "paginas_por_prestamo"]])
````

Obtener el resumen de la columna `paginas` con **`describe`**, redondeado a un decimal. A partir de la salida, indicar si hay algún libro mucho más largo que el resto.

> **Ayuda:** [`Series.describe`](https://pandas.pydata.org/docs/reference/api/pandas.Series.describe.html). El significado de cada fila está en el apartado 3 del anexo 3.

**Datos de partida.** Páginas aproximadas de ocho libros, entre ellos *El segundo sexo*, de [Simone de Beauvoir](https://es.wikipedia.org/wiki/Simone_de_Beauvoir).

````python
import pandas as pd

paginas = pd.Series([160, 224, 288, 912, 320, 280, 432, 296],
                    index=["Una habitación propia", "La señora Dalloway",
                           "La campana de cristal", "El segundo sexo",
                           "El género en disputa", "Frankenstein",
                           "Orgullo y prejuicio", "Nada"], name="paginas")
````

### Ejercicio 12. Resumen por grupos

Calcular el **número de libros**, la **media** y la **mediana** de préstamos de cada género, con la media redondeada a un decimal.

> **Ayuda:** [`DataFrame.groupby`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.groupby.html) seguido de [`agg`](https://pandas.pydata.org/docs/reference/api/pandas.core.groupby.DataFrameGroupBy.agg.html) con una lista de nombres de funciones. Repaso en el apartado 5 del anexo 3.

**Datos de partida.** Género y préstamos de ocho libros (datos ilustrativos).

````python
import pandas as pd

libros = pd.DataFrame({
    "genero": ["ensayo", "novela", "novela", "poesía", "ensayo",
               "novela", "ensayo", "novela"],
    "prestamos": [34, 21, 29, 8, 15, 41, 9, 26],
})
````

**Resultado esperado.**

```text
        count  mean  median
genero                     
ensayo      3  19.3    15.0
novela      4  29.2    27.5
poesía      1   8.0     8.0
```

````python
# Escribe aquí tu solución y compara el resultado con el esperado
````

#### Solución del ejercicio 12

````python
resumen = libros.groupby("genero")["prestamos"].agg(["count", "mean", "median"])
print(resumen.round(1))
````

`groupby` parte la tabla en un trozo por género y `agg` aplica las tres funciones a cada trozo. En los ensayos, la media (19,3) queda muy por encima de la mediana (15) por culpa de un solo ensayo muy prestado (34). La poesía, con un único libro, no permite comparar nada.
