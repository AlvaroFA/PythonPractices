### Ejercicio 1. Media y mediana ante un valor extremo

Calcular la **media** y la **mediana** de las duraciones de seis películas. Repetir el cálculo **sin la película más larga** y comparar cuánto cambia cada medida.

> **Ayuda:** los métodos [`Series.mean`](https://pandas.pydata.org/docs/reference/api/pandas.Series.mean.html), [`Series.median`](https://pandas.pydata.org/docs/reference/api/pandas.Series.median.html) y [`Series.drop`](https://pandas.pydata.org/docs/reference/api/pandas.Series.drop.html). Repaso en los apartados 2.1 y 2.2 del anexo 2.

**Datos de partida.** Duración, en minutos, de seis películas dirigidas por mujeres: *Cleo de 5 a 7*, de [Agnès Varda](https://es.wikipedia.org/wiki/Agn%C3%A8s_Varda); *Las margaritas*, de [Věra Chytilová](https://es.wikipedia.org/wiki/V%C4%9Bra_Chytilov%C3%A1); *El autoestopista*, de [Ida Lupino](https://es.wikipedia.org/wiki/Ida_Lupino); *Baila, muchacha, baila*, de [Dorothy Arzner](https://es.wikipedia.org/wiki/Dorothy_Arzner); *La ascensión*, de [Larisa Shepitko](https://es.wikipedia.org/wiki/Larisa_Shepitko), y *Jeanne Dielman*, de [Chantal Akerman](https://es.wikipedia.org/wiki/Chantal_Akerman).

````python
import pandas as pd

duracion = pd.Series(
    [90, 76, 71, 90, 111, 201],
    index=["Cleo de 5 a 7", "Las margaritas", "El autoestopista",
           "Baila, muchacha, baila", "La ascensión", "Jeanne Dielman"],
)
````


````python
# Escribe aquí tu solución y compara el resultado con el esperado
````

#### Solución del ejercicio 1

````python
print("Con las seis:   media", round(duracion.mean(), 1), "  mediana", duracion.median())
sin_larga = duracion.drop("Jeanne Dielman")
print("Sin la más larga: media", round(sin_larga.mean(), 1), "  mediana", sin_larga.median())
````

Los 201 minutos de *Jeanne Dielman* suben la media casi 19 minutos, hasta un valor (106,5) que no describe a ninguna de las otras cinco películas. La mediana no se mueve: con seis valores es la media de los dos centrales (90 y 90), y con cinco, el central (90). La mediana solo depende del **orden** de los datos, no de lo lejos que esté el último.

**Datos de partida.** Edad a la que publicaron su primera novela [Mary Shelley](https://es.wikipedia.org/wiki/Mary_Shelley), [Jane Austen](https://es.wikipedia.org/wiki/Jane_Austen), [Emily Brontë](https://es.wikipedia.org/wiki/Emily_Bront%C3%AB), [Virginia Woolf](https://es.wikipedia.org/wiki/Virginia_Woolf), [Simone de Beauvoir](https://es.wikipedia.org/wiki/Simone_de_Beauvoir) y [Sylvia Plath](https://es.wikipedia.org/wiki/Sylvia_Plath).

### Ejercicio 3. Cuartiles y rango intercuartílico

Calcular el **primer cuartil**, la **mediana** y el **tercer cuartil**, el **rango intercuartílico** (IQR) y el umbral superior de la regla $Q_3 + 1{,}5\,\text{IQR}$. Mostrar los valores que superan ese umbral.

> **Ayuda:** el método [`Series.quantile`](https://pandas.pydata.org/docs/reference/api/pandas.Series.quantile.html), que admite una lista de proporciones. Repaso en el apartado 3 del anexo 2.

**Datos de partida.** Minutos diarios de lectura de ocho estudiantes de un club de lectura (datos ilustrativos).

````python
import pandas as pd

lectura = pd.Series([10, 15, 20, 20, 25, 30, 40, 90])
````

**Resultado esperado.**

```text
Q1: 18.75   mediana: 22.5   Q3: 32.5
IQR: 13.75   umbral superior: 53.125
Por encima del umbral: [90]
```

````python
# Escribe aquí tu solución y compara el resultado con el esperado
````

#### Solución del ejercicio 3

````python
q1, mediana, q3 = lectura.quantile([0.25, 0.5, 0.75])
iqr = q3 - q1
umbral = q3 + 1.5 * iqr
print("Q1:", q1, "  mediana:", mediana, "  Q3:", q3)
print("IQR:", iqr, "  umbral superior:", umbral)
print("Por encima del umbral:", lectura[lectura > umbral].tolist())
````

El 50 % central de los datos está entre 18,75 y 32,5 minutos, una franja de 13,75 minutos. Los 90 minutos quedan muy por encima del umbral (53,125) y se señalan como **atípicos**. pandas interpola entre los dos valores más próximos cuando el cuartil cae entre dos observaciones; por eso Q1 vale 18,75 aunque nadie lea exactamente ese tiempo.

### Ejercicio 4. Puntuación z para comparar notas

Una alumna ha sacado un 7 en Estadística y un 8 en Python. Calcular su **puntuación z** en cada asignatura, con la media y la desviación típica muestral de su grupo, y decidir en cuál destaca más.

> **Ayuda:** la fórmula $z = (x - \bar{x}) / s$, con [`Series.mean`](https://pandas.pydata.org/docs/reference/api/pandas.Series.mean.html) y [`Series.std`](https://pandas.pydata.org/docs/reference/api/pandas.Series.std.html). Repaso en el apartado 6 del anexo 2.

**Datos de partida.** Notas de los seis estudiantes del grupo en las dos asignaturas.

````python
import pandas as pd

notas = pd.DataFrame({
    "estadistica": [3, 4, 5, 5, 6, 7],
    "python": [4, 6, 7, 8, 9, 8],
})
````

**Resultado esperado.**

```text
z en Estadística: 1.414
z en Python:      0.559
```

````python
# Escribe aquí tu solución y compara el resultado con el esperado
````

#### Solución del ejercicio 4

````python
z_estadistica = (7 - notas["estadistica"].mean()) / notas["estadistica"].std()
z_python = (8 - notas["python"].mean()) / notas["python"].std()
print("z en Estadística:", round(z_estadistica, 3))
print("z en Python:     ", round(z_python, 3))
````

El 8 de Python es la nota más alta en valor absoluto, pero en Python el grupo saca de media un 7, de modo que solo queda 0,56 desviaciones típicas por encima. El 7 de Estadística está 1,41 desviaciones por encima de una media de 5: **en Estadística destaca más**. La puntuación z quita las unidades y permite comparar valores de escalas distintas.

### Ejercicio 6. Covarianza a mano y con pandas

Calcular la **covarianza muestral** entre las horas de estudio y la nota **a mano**, con la fórmula $s_{{xy}} = \frac{{1}}{{n-1}}\sum (x_i - \bar{{x}})(y_i - \bar{{y}})$, y comprobar el resultado con pandas.

> **Ayuda:** el método [`DataFrame.cov`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.cov.html), que devuelve la matriz de covarianzas. Repaso en el apartado 3.

**Datos de partida.** Horas de estudio y nota de cinco estudiantes.

````python
import pandas as pd

estudio = pd.DataFrame({"horas": [1, 2, 3, 4, 5], "nota": [4, 5, 5, 7, 9]})
````

**Resultado esperado.**

```text
Covarianza a mano: 3.0
       horas  nota
horas    2.5   3.0
nota     3.0   4.0
```

````python
# Escribe aquí tu solución y compara el resultado con el esperado
````

#### Solución del ejercicio 6

````python
dx = estudio["horas"] - estudio["horas"].mean()
dy = estudio["nota"] - estudio["nota"].mean()
a_mano = (dx * dy).sum() / (len(estudio) - 1)
print("Covarianza a mano:", a_mano)
print(estudio.cov())
````

Los productos de las desviaciones son 4, 1, 0, 1 y 6: casi todos positivos, porque quien estudia más horas que la media suele sacar más nota que la media. Su suma (12) entre $n - 1 = 4$ da 3. En la matriz, la covarianza aparece fuera de la diagonal y las **varianzas** de cada variable, en la diagonal.

### Ejercicio 7. Cambio de unidades: covarianza frente a correlación

Calcular la covarianza y la correlación de Pearson entre la estatura y el peso con la estatura **en metros** y **en centímetros**. Comparar cómo cambia cada una.

> **Ayuda:** [`Series.cov`](https://pandas.pydata.org/docs/reference/api/pandas.Series.cov.html) y [`Series.corr`](https://pandas.pydata.org/docs/reference/api/pandas.Series.corr.html). Repaso en los apartados 3 y 4.

**Datos de partida.** Estatura y peso de cinco personas.

````python
import pandas as pd

personas = pd.DataFrame({
    "estatura_m": [1.60, 1.65, 1.70, 1.75, 1.80],
    "peso_kg": [55, 60, 63, 70, 72],
})
personas["estatura_cm"] = personas["estatura_m"] * 100
````

**Resultado esperado.**

```text
estatura_m   covarianza =   0.550   Pearson = 0.9888
estatura_cm  covarianza =  55.000   Pearson = 0.9888
```

````python
# Escribe aquí tu solución y compara el resultado con el esperado
````

#### Solución del ejercicio 7

````python
for columna in ["estatura_m", "estatura_cm"]:
    cov = personas[columna].cov(personas["peso_kg"])
    r = personas[columna].corr(personas["peso_kg"])
    print(f"{columna:12} covarianza = {cov:7.3f}   Pearson = {r:.4f}")
````

Al pasar a centímetros, la covarianza se multiplica por 100, aunque las personas y su relación son exactamente las mismas. El coeficiente de Pearson no cambia, porque divide la covarianza entre las dos desviaciones típicas y las unidades se cancelan. Por eso **la covarianza indica el sentido de la relación, pero no su fuerza**.

### Ejercicio 8. Coeficiente de Pearson con pandas y SciPy

Calcular el **coeficiente de Pearson** entre el periodo y la luminosidad de las cefeidas con pandas y con SciPy. Mostrar también el **p-valor** que devuelve SciPy.

> **Ayuda:** [`Series.corr`](https://pandas.pydata.org/docs/reference/api/pandas.Series.corr.html) y [`scipy.stats.pearsonr`](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.pearsonr.html), que devuelve un objeto con los atributos `statistic` y `pvalue`. Repaso en el apartado 4.1.

**Datos de partida.** Las seis estrellas cefeidas del ejemplo de la presentación de la unidad, inspirado en el trabajo de [Henrietta Swan Leavitt](https://es.wikipedia.org/wiki/Henrietta_Swan_Leavitt).

````python
import pandas as pd
from scipy import stats

cefeidas = pd.DataFrame({
    "periodo_dias": [4.3, 5.9, 11.2, 16.5, 30.1, 45.0],
    "luminosidad": [350, 480, 1100, 1850, 4200, 7500],
})
````

**Resultado esperado.**

```text
Pearson con pandas: 0.9937
Pearson con SciPy:  0.9937
p-valor:            5.94e-05
```

````python
# Escribe aquí tu solución y compara el resultado con el esperado
````

#### Solución del ejercicio 8

````python
r_pandas = cefeidas["periodo_dias"].corr(cefeidas["luminosidad"])
resultado = stats.pearsonr(cefeidas["periodo_dias"], cefeidas["luminosidad"])
print("Pearson con pandas:", round(r_pandas, 4))
print("Pearson con SciPy: ", round(resultado.statistic, 4))
print("p-valor:           ", f"{resultado.pvalue:.2e}")
````

Los dos coeficientes coinciden: hay una **relación lineal positiva muy fuerte**. El p-valor, muy por debajo de 0,001, indica que una correlación así sería muy improbable si en realidad no hubiera relación. Con solo seis datos lo prudente es no ir más allá: el p-valor dice que la relación existe, y el coeficiente, lo fuerte que es.

### Ejercicio 9. Spearman como Pearson de los rangos

Con las mismas cefeidas, calcular el **coeficiente de Spearman**. Después, **sustituir cada valor por su rango** (su puesto en la ordenación) y calcular Pearson sobre los rangos. Comprobar que coinciden.

> **Ayuda:** [`Series.corr`](https://pandas.pydata.org/docs/reference/api/pandas.Series.corr.html) con `method="spearman"` y [`Series.rank`](https://pandas.pydata.org/docs/reference/api/pandas.Series.rank.html). Repaso en el apartado 5.

**Datos de partida.** Las seis cefeidas del ejercicio anterior.

````python
import pandas as pd

cefeidas = pd.DataFrame({
    "periodo_dias": [4.3, 5.9, 11.2, 16.5, 30.1, 45.0],
    "luminosidad": [350, 480, 1100, 1850, 4200, 7500],
})
````

**Resultado esperado.**

```text
   periodo_dias  luminosidad
0           1.0          1.0
1           2.0          2.0
2           3.0          3.0
3           4.0          4.0
4           5.0          5.0
5           6.0          6.0
Spearman:             1.0
Pearson de los rangos: 1.0
```

````python
# Escribe aquí tu solución y compara el resultado con el esperado
````

#### Solución del ejercicio 9

````python
rho = cefeidas["periodo_dias"].corr(cefeidas["luminosidad"], method="spearman")
rangos = cefeidas.rank()
print(rangos)
print("Spearman:            ", rho)
print("Pearson de los rangos:", rangos["periodo_dias"].corr(rangos["luminosidad"]))
````

La luminosidad crece **siempre** que crece el periodo, aunque no en línea recta: la relación es **monótona**. Por eso los dos rangos coinciden puesto a puesto (1, 2, ..., 6) y Spearman vale exactamente 1, mientras que Pearson se quedaba algo por debajo. Spearman no mira cuánto crece una variable, solo si crece.

### Ejercicio 11. Tabla de contingencia

Construir la **tabla de contingencia** entre la franja de edad y el formato de lectura preferido, con los **totales** de filas y columnas. Después, calcular la **proporción de cada formato dentro de cada franja de edad**.

> **Ayuda:** [`pd.crosstab`](https://pandas.pydata.org/docs/reference/api/pandas.crosstab.html), con los parámetros `margins` y `normalize`. Repaso en el apartado 7.2.

**Datos de partida.** Franja de edad y formato preferido de diez socias de un club de lectura.

````python
import pandas as pd

socias = pd.DataFrame({
    "edad": ["menos de 30", "menos de 30", "30 o más", "menos de 30", "30 o más",
             "30 o más", "menos de 30", "30 o más", "30 o más", "menos de 30"],
    "formato": ["digital", "digital", "papel", "papel", "papel",
                "papel", "digital", "digital", "papel", "digital"],
})
````

### Ejercicio 12. Contraste chi-cuadrado

Aplicar el **contraste chi-cuadrado de independencia** a la tabla de contingencia. Mostrar el estadístico, el p-valor, los grados de libertad y la **tabla de frecuencias esperadas** si las dos variables fueran independientes.

> **Ayuda:** [`scipy.stats.chi2_contingency`](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.chi2_contingency.html), que devuelve un objeto con los atributos `statistic`, `pvalue`, `dof` y `expected_freq`. Repaso en el apartado 7 y, paso a paso, en el anexo 1.

**Datos de partida.** Formato de lectura preferido por 120 personas, según su franja de edad.

````python
import pandas as pd
from scipy import stats

tabla = pd.DataFrame(
    [[20, 25, 15], [35, 15, 10]],
    index=["menos de 30", "30 o más"],
    columns=["papel", "digital", "audiolibro"],
)
````

**Resultado esperado.**

```text
Chi-cuadrado: 7.591
p-valor: 0.0225
Grados de libertad: 2
             papel  digital  audiolibro
menos de 30   27.5     20.0        12.5
30 o más      27.5     20.0        12.5
```

````python
# Escribe aquí tu solución y compara el resultado con el esperado
````

#### Solución del ejercicio 12

````python
resultado = stats.chi2_contingency(tabla)
print("Chi-cuadrado:", round(resultado.statistic, 3))
print("p-valor:", round(resultado.pvalue, 4))
print("Grados de libertad:", resultado.dof)
esperadas = pd.DataFrame(resultado.expected_freq, index=tabla.index, columns=tabla.columns)
print(esperadas)
````

Si la edad no influyera, cada franja repartiría sus 60 personas como el total: 27,5 en papel, 20 en digital y 12,5 en audiolibro. Lo observado se aleja de eso (35 frente a 27,5 en papel en el grupo de 30 o más) y el p-valor, 0,023, queda por debajo de 0,05: **se rechaza la independencia**. Los grados de libertad son $(2 - 1)(3 - 1) = 2$.

### Ejercicio 13. Intensidad de la asociación: V de Cramér

Medir la **intensidad** de la asociación de la tabla anterior con la **V de Cramér**, calculada con SciPy y con la fórmula $V = \sqrt{{\chi^2 / (n\,(k-1))}}$, donde $k$ es el menor entre el número de filas y de columnas.

> **Ayuda:** [`scipy.stats.contingency.association`](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.contingency.association.html), con `method="cramer"`. Repaso en el apartado 7.4.

**Datos de partida.** La misma tabla del ejercicio anterior.

````python
import numpy as np
import pandas as pd
from scipy import stats

tabla = pd.DataFrame(
    [[20, 25, 15], [35, 15, 10]],
    index=["menos de 30", "30 o más"],
    columns=["papel", "digital", "audiolibro"],
)
````

**Resultado esperado.**

```text
V de Cramér con SciPy:   0.252
V de Cramér con fórmula: 0.252
```

````python
# Escribe aquí tu solución y compara el resultado con el esperado
````

#### Solución del ejercicio 13

````python
v_scipy = stats.contingency.association(tabla, method="cramer")
chi2 = stats.chi2_contingency(tabla).statistic
n = tabla.to_numpy().sum()
k = min(tabla.shape)
v_formula = np.sqrt(chi2 / (n * (k - 1)))
print("V de Cramér con SciPy:  ", round(v_scipy, 3))
print("V de Cramér con fórmula:", round(v_formula, 3))
````

Las dos vías dan 0,252, una **asociación débil** según la regla orientativa del apartado 7.4 (entre 0,1 y 0,3).
